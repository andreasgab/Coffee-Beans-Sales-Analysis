```python
import pandas as pd

# pd.set_option('display.max_rows', None)
# pd.set_option('display.max_columns', None)
# pd.reset_option('display.max_rows')
# pd.reset_option('display.max_columns')

df1 = pd.read_excel(r"C:\Users\hyste\OneDrive\Έγγραφα\Python\My Projects\Github Repo\Coffee-Beans-Sales-Analysis\data\Raw Data.xlsx",
                    sheet_name='orders',
                    header=0,)
df2 = pd.read_excel(r"C:\Users\hyste\OneDrive\Έγγραφα\Python\My Projects\Github Repo\Coffee-Beans-Sales-Analysis\data\Raw Data.xlsx",
                    sheet_name='customers',
                    header=0,)
df3 = pd.read_excel(r"C:\Users\hyste\OneDrive\Έγγραφα\Python\My Projects\Github Repo\Coffee-Beans-Sales-Analysis\data\Raw Data.xlsx",
                    sheet_name='products',
                    header=0,)

df1=df1[['Order ID','Order Date','Customer ID','Product ID','Quantity']]

display(df1,df2,df3)

combined = pd.merge(df1, df2, on='Customer ID', how='left')
combined = pd.merge(combined, df3, on='Product ID', how='left')
combined = combined.sort_values(by='Order Date')

combined['Day'] = combined['Order Date'].dt.day
combined['Weekday'] = combined['Order Date'].dt.day_name()
combined['Month'] = combined['Order Date'].dt.month
combined['Year'] = combined['Order Date'].dt.year

display(combined)
# combined.dtypes
```

```python
combined['Coffee Type'] = combined['Coffee Type'].str.replace('Ara', 'Arabica')
combined['Coffee Type'] = combined['Coffee Type'].str.replace('Rob', 'Robusta')
combined['Coffee Type'] = combined['Coffee Type'].str.replace('Exc', 'Excelsa')
combined['Coffee Type'] = combined['Coffee Type'].str.replace('Lib', 'Liberica')

combined['Roast Type'] = combined['Roast Type'].str.replace('L', 'Light Roast')
combined['Roast Type'] = combined['Roast Type'].str.replace('M', 'Medium Roast')
combined['Roast Type'] = combined['Roast Type'].str.replace('D', 'Dark Roast')

# combined.replace({  'Roast Type': {'L': 'Light', 'M': 'Medium', 'D': 'Dark'},
#                     'Coffee Type': {'Ara': 'Arabica', 'Rob': 'Robusta', 'Exc': 'Excelsa', 'Lib': 'Liberica'}}, inplace=True)
```

```python
combined = combined.drop(columns = 'Price per 100g')

combined['Revenue'] = combined['Unit Price'] * combined['Quantity']

cols = list(combined.columns)
q = cols.pop(cols.index('Quantity'))
cols.insert(cols.index('Roast Type') + 1, q)
r = cols.pop(cols.index('Revenue'))
cols.insert(cols.index('Unit Price') + 1, r)

combined = combined[cols]

combined
```
